# Сравнение скорости: list comprehension vs numpy
Создаём массив из 100,000 случайных чисел.

```python
import numpy as np
import random

# Параметры
N = 100_000

print("List comprehension:")
%timeit [random.random() for _ in range(N)]

print("\nNumPy:")
%timeit np.random.random(N)
```

# Работа с CSV и визуализацией
Генерируем тестовый CSV, загружаем его, выводим таблицу и строим график.

```python
import pandas as pd
import numpy as np

# Создадим простой DataFrame и сохраним в CSV
df = pd.DataFrame({
    'x': np.arange(0, 10, 0.1),
    'y': np.sin(np.arange(0, 10, 0.1))
})
df.to_csv('test_data.csv', index=False)
print("CSV сохранён")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Загружаем CSV
data = pd.read_csv('test_data.csv')

# Выводим первые 5 строк
print("Первые строки:")
display(data.head())

# Строим график
plt.figure(figsize=(8,4))
plt.plot(data['x'], data['y'], marker='o', linestyle='-', alpha=0.7)
plt.title('Синусоида')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
